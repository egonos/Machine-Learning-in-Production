## The Supervised Learning Problem

In the pursuit of supervised machine learning (ML), our goal is to predict the response variable $y$ as best we can, given the predictor matrix $X$ (more technically, we model the conditional distribution $P(y \mid X)$). To do that, we give the $(X, y)$ pair to the model. The model automatically comes up with equations, decision rules, etc. to make predictions $\hat{y}$ from $X$ that are aligned with the true label $y$. We call this "model fitting".
