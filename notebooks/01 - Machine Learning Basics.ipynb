## 1. The Supervised Learning Problem

- In the pursuit of supervised machine learning (ML), our goal is to predict response variable $y$ as best as we can given predictor matrix $X$ (more technically maximize y|X). To do that, we fit 
- Target functional: conditional expectation vs conditional quantile
- Hypothesis space, parametric vs non-parametric models
- Estimation error vs approximation error vs irreducible noise

## 2. Loss Functions and What They Actually Estimate

- Squared error estimates the mean, absolute error estimates the median
- Pinball loss estimates a quantile
- Log loss and Brier score estimate a probability
- Choosing the loss from the decision, not from habit

## 3. Bias-Variance Decomposition by Simulation

- A controlled data generating process with known truth
- Repeated sampling to measure bias, variance and noise separately
- Model complexity sweep and the decomposition plot
