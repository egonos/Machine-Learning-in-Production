# Data Drift and Concept Drift

> Block 2 - Time and Streaming Data

Detecting, explaining and responding to a world that stops matching the training set.

## 1. A Taxonomy of Shift

- Covariate shift: P(X) changes
- Prior / label shift: P(y) changes
- Concept drift: P(y|X) changes
- Virtual drift: shift that does not hurt performance

## 2. Drift Patterns

- Sudden, gradual, incremental, recurring
- Seasonality mistaken for drift
- Simulating each pattern with a controlled DGP

## 3. Distribution-Based Detection

- Population Stability Index (PSI) and its thresholds
- Kolmogorov-Smirnov and Chi-square tests
- Wasserstein distance and Jensen-Shannon divergence
- Maximum Mean Discrepancy (MMD)
- Categorical and high-cardinality features

## 4. The Multiple Testing Problem in Drift Monitoring

- With 200 features you will always find drift
- Correction strategies and alert budgets

## 5. Model-Based Detection

- Domain classifier / classifier two-sample test
- Adversarial validation AUC as a drift magnitude
- Reconstruction error and density ratio estimation

## 6. Performance-Based Drift Detectors

- DDM and EDDM
- ADWIN adaptive windowing
- Page-Hinkley and CUSUM

## 7. Estimating Performance Without Labels

- Importance weighting under covariate shift
- Confidence-based estimation (ATC, DoC)
- Limitations and failure modes

## 8. Explaining Drift

- Which features moved, and how much do they matter?
- SHAP on the domain classifier
- Separating drift in inputs from drift in the relationship

## 9. Prediction Drift and Calibration Decay

- Monitoring the output distribution
- Recalibration as a first response

## 10. Responses to Drift

- Retraining windows: sliding, expanding, weighted
- Importance weighting of old data
- Online and incremental learning
- Adaptive ensembles and model pools

## 11. When Not to Retrain

- Cost, stability and regression risk
- Feedback loops that amplify drift
