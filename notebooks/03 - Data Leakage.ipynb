# Data Leakage

> Block 1 - Validation and Evaluation Protocol

The failure mode that makes a model look excellent offline and useless in production.

## 1. A Taxonomy of Leakage

- Target leakage, train/test contamination, temporal leakage
- Group leakage, preprocessing leakage, duplicate leakage

## 2. Target Leakage

- Proxies of the target hidden in the feature set
- Post-outcome features that only exist after the decision
- Demo: a 0.99 AUC that collapses in production

## 3. Preprocessing Leakage

- Scaling, imputation and PCA fitted before the split
- Feature selection performed on the full dataset
- Quantifying the optimistic bias

## 4. Resampling Leakage

- SMOTE and oversampling applied before splitting
- Duplicated synthetic neighbours across folds

## 5. Target / Mean Encoding Done Right

- Out-of-fold encoding and smoothing
- Nested encoding inside cross validation
- Leave-one-out encoding and its failure mode

## 6. Temporal and Group Leakage

- Using the future to predict the past
- Overlapping label windows
- Bridge to notebooks 02 and 07

## 7. Duplicate and Near-Duplicate Records

- Exact duplicates across splits
- Near-duplicates and entity resolution

## 8. Detecting Leakage in Practice

- Suspiciously high scores and dominant feature importance
- Performance decay over time
- Adversarial validation between train and test
- Ablation: drop the suspect feature and re-measure

## 9. Structural Defenses: Pipeline and ColumnTransformer

- Making leakage impossible instead of merely unlikely
- Custom transformers that respect fit/transform boundaries
